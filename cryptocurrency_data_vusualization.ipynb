```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Cryptocurrency Market Analysis Using CoinGecko API\clean_crypto_data.csv")
df
```

```python
df.drop(columns=('Unnamed: 0'),inplace=True)
df
```

```python
#1 Top 10 Cryptocurrencies by Market Cap — Matplotlib
top10 = df.nlargest(10, "Market_Cap")

plt.figure(figsize=(12, 6))

# Different color for each bar
colors = [
    "orange", "royalblue", "green", "gold", "purple",
    "red", "cyan", "pink", "brown", "teal"
]

plt.bar(
    top10["Name"],
    top10["Market_Cap"],
    color=colors
)

plt.title(
    "Top 10 Cryptocurrencies by Market Cap",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Cryptocurrency", fontsize=12)
plt.ylabel("Market Cap", fontsize=12)

plt.xticks(rotation=45)

# Add grid
plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

```python
#2  Market Cap vs Total Volume — Seaborn Scatter Plot
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Market_Cap",
    y="Total_Volume"
)

plt.title("Market Cap vs Total Trading Volume")
plt.xlabel("Market Cap")
plt.ylabel("Total Volume")

plt.tight_layout()
plt.show()
```

```python
#3 Distribution of 24-Hour Price Change
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Price_Change_Percentage_24H",
    bins=20,
    kde=True
)

plt.title("Distribution of 24-Hour Price Change")
plt.xlabel("24-Hour Price Change (%)")
plt.ylabel("Number of Cryptocurrencies")

plt.tight_layout()
plt.show()
```

```python
#4 Top 10 Cryptocurrencies by 24-Hour Price Change

top_change = df.nlargest(
    10,
    "Price_Change_Percentage_24H"
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_change,
    x="Name",
    y="Price_Change_Percentage_24H",
    palette="husl"
)

plt.title(
    "Top 10 Cryptocurrencies by 24-Hour Price Change",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Cryptocurrency",
    fontsize=12
)

plt.ylabel(
    "Price Change (%)",
    fontsize=12
)

plt.xticks(rotation=45)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

```python
# 5. Count Plot — Rating
sns.countplot(data=df, x="Name")

plt.title("Number of Cryptocurrencies")
plt.xlabel("Cryptocurrency")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```
